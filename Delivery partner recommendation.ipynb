```python
def get_distance_class(distance):
    if distance <= 5:
        return '0-5'
    elif distance <= 10:
        return '5-10'
    elif distance <= 15:
        return '10-15'
    else:
        return '15>'
```

```python
# this is for sorting based on driver score; mostly for backend

def recommendation(weather, traffic, festival, city, distance):

    # Select scores based on input
    weather_score = weather_performance[
        weather_performance['Weather_conditions'] == weather
    ].copy()

    traffic_score = traffic_performance[
        traffic_performance['Road_traffic_density'] == traffic
    ].copy()

    festival_score = festival_performance[
        festival_performance['Festival'] == festival
    ].copy()

    city_score = city_performance[
        city_performance['City'] == city
    ].copy()

    distance_class = get_distance_class(distance)

    distance_score = distance_performance[
        distance_performance['distance_class'] == distance_class
    ].copy()

    # Rename average-time columns
    weather_score = weather_score.rename(
        columns={'avg_time': 'weather_avg_time'}
    )

    traffic_score = traffic_score.rename(
        columns={'avg_time': 'traffic_avg_time'}
    )

    festival_score = festival_score.rename(
        columns={'avg_time': 'festival_avg_time'}
    )

    city_score = city_score.rename(
        columns={'avg_time': 'city_avg_time'}
    )

    distance_score = distance_score.rename(
        columns={'avg_time': 'distance_avg_time'}
    )

    # Merge all scores
    recommendation_df = dp.merge(
        weather_score,
        on='Delivery_person_ID',
        how='left'
    )

    recommendation_df = recommendation_df.merge(
        traffic_score,
        on='Delivery_person_ID',
        how='left'
    )

    recommendation_df = recommendation_df.merge(
        festival_score,
        on='Delivery_person_ID',
        how='left'
    )

    recommendation_df = recommendation_df.merge(
        city_score,
        on='Delivery_person_ID',
        how='left'
    )

    recommendation_df = recommendation_df.merge(
        distance_score,
        on='Delivery_person_ID',
        how='left'
    )

    # Condition-specific average delivery time
    score_columns = [
        'weather_avg_time',
        'traffic_avg_time',
        'festival_avg_time',
        'city_avg_time',
        'distance_avg_time'
    ]

    recommendation_df['total_avg_time'] = recommendation_df[
        score_columns
    ].mean(axis=1)

    # Count missing scores
    recommendation_df['nan_count'] = recommendation_df[
        score_columns
    ].isna().sum(axis=1)

    # Normalize rating
    recommendation_df['rating_score'] = (
        recommendation_df['avg_deliverer_rating']
        - recommendation_df['avg_deliverer_rating'].min()
    ) / (
        recommendation_df['avg_deliverer_rating'].max()
        - recommendation_df['avg_deliverer_rating'].min()
    )

    # Normalize orders
    recommendation_df['orders_score'] = (
        recommendation_df['orders_completed']
        - recommendation_df['orders_completed'].min()
    ) / (
        recommendation_df['orders_completed'].max()
        - recommendation_df['orders_completed'].min()
    )

    # Normalize delivery time
    recommendation_df['time_score'] = (
        recommendation_df['total_avg_time'].max()
        - recommendation_df['total_avg_time']
    ) / (
        recommendation_df['total_avg_time'].max()
        - recommendation_df['total_avg_time'].min()
    )

    # Combined score
    recommendation_df['driver_score'] = (
        0.30 * recommendation_df['rating_score']
        + 0.20 * recommendation_df['orders_score']
        + 0.50 * recommendation_df['time_score']
    )

    # Sort first by least NaNs then by higher combined driver score
    recommendation_df = recommendation_df.sort_values(
        ['nan_count', 'driver_score'],
        ascending=[True, False]
    )

    # Keep only 4 unique delivery persons
    recommendation_df = recommendation_df.drop_duplicates(
        'Delivery_person_ID'
    ).head(4)

    return recommendation_df
```

```python
recommendation(
    weather='Sunny',
    traffic='Low',
    festival='Yes',
    city='Urban',
    distance= 4 
)

```

```python
def customer_recommendation(recommendation_df):

    df = recommendation_df.drop_duplicates(
        'Delivery_person_ID'
    ).copy()

    selected = []

    # with highest rating
    option_1 = df.sort_values(
        'avg_deliverer_rating',
        ascending=False
    ).iloc[0]

    selected.append(option_1['Delivery_person_ID'])

    # fastest driver
    option_2 = df[
        ~df['Delivery_person_ID'].isin(selected)
    ].sort_values(
        'total_avg_time',
        ascending=True
    ).iloc[0]

    selected.append(option_2['Delivery_person_ID'])

    # highest order counts
    option_3 = df[
        ~df['Delivery_person_ID'].isin(selected)
    ].sort_values(
        'orders_completed',
        ascending=False
    ).iloc[0]

    selected.append(option_3['Delivery_person_ID'])

    # balanced queen
    remaining = df[
        ~df['Delivery_person_ID'].isin(selected)
    ].copy()

    # normalize the three factors
    remaining['rating_score'] = (
        remaining['avg_deliverer_rating']
        - df['avg_deliverer_rating'].min()
    ) / (
        df['avg_deliverer_rating'].max()
        - df['avg_deliverer_rating'].min()
    )

    remaining['order_score'] = (
        remaining['orders_completed']
        - df['orders_completed'].min()
    ) / (
        df['orders_completed'].max()
        - df['orders_completed'].min()
    )

    remaining['time_score'] = (
        df['total_avg_time'].max()
        - remaining['total_avg_time']
    ) / (
        df['total_avg_time'].max()
        - df['total_avg_time'].min()
    )

    # equal importance to all three
    remaining['balance_score'] = (
        0.33 * remaining['rating_score']
        + 0.33 * remaining['order_score']
        + 0.34 * remaining['time_score']
    )

    option_4 = remaining.sort_values(
        'balance_score',
        ascending=False
    ).iloc[0]

    # Combining all 4 options
    customer_df = pd.DataFrame([
        option_1,
        option_2,
        option_3,
        option_4
    ])

    # keeping only what the customer needs to see
    customer_df = customer_df[
        [
            'Delivery_person_ID',
            'avg_deliverer_rating',
            'orders_completed',
            'total_avg_time'
        ]
    ]

    # Renaming for aesthetics hehe
    customer_df = customer_df.rename(
        columns={
            'Delivery_person_ID': 'Delivery Person',
            'avg_deliverer_rating': 'Rating',
            'orders_completed': 'Order Count',
            'total_avg_time': 'Avg Delivery Time'
        }
    )

    return customer_df
```

```python
detailed_result = recommendation(
    weather='Sunny',
    traffic='High',
    festival='Yes',
    city='Urban',
    distance=16
)

customer_result = customer_recommendation(detailed_result)

customer_result
```
